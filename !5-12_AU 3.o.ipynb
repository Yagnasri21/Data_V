```python
# plot distribution of generated power
plt.figure(figsize=(8, 4))
plt.hist(df['generated_power_kw'], bins=30, edgecolor='black', alpha=0.7, density=False)
plt.xlabel('generated_power_kw')
plt.ylabel('Count')
plt.title('Distribution of generated_power_kw')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# scatter plot for each feature against the target audience
plt.figure(figsize=(15,30))
for num,column in enumerate(df.columns):
    if column != 'generated_power_kw':
        plt.subplot(7, 3, num+1)
        plt.scatter(df[column],df['generated_power_kw'])
        plt.xlabel(column)
        plt.ylabel('Generated Power (KW)')
        plt.title(f'{column} vs Generated Power')
plt.tight_layout()
plt.show()
```

```python
# plot correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
plt.title('Correlation Matrix')
plt.show()
```

```python
# plot correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
plt.title('Correlation Matrix')
plt.show()
```

```python

plt.figure(figsize=(10, 8))
im = plt.imshow(correlation_matrix.values, cmap='coolwarm', aspect='auto')

plt.colorbar(im)

for i in range(correlation_matrix.shape[0]):
    for j in range(correlation_matrix.shape[1]):
        text = plt.text(j, i, f'{correlation_matrix.iloc[i, j]:.2f}',
                       ha="center", va="center", color="black", fontsize=10)

plt.title('Correlation Matrix')
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45, ha='right')
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.tight_layout()
plt.show()
```

```python
# Plot correlation matrix (no row/column labels)
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()

im = plt.imshow(correlation_matrix.values, cmap='coolwarm', aspect='auto')

# Add annotations only (no ticks/labels)
for i in range(correlation_matrix.shape[0]):
    for j in range(correlation_matrix.shape[1]):
        plt.text(j, i, f'{correlation_matrix.iloc[i, j]:.2f}',
                ha="center", va="center", color="black", fontsize=11, weight='bold')

plt.colorbar(im)
plt.title('Correlation Matrix')
plt.axis('off')  # Remove all axis labels and ticks
plt.tight_layout()
plt.show()
```

```python
# plot the distribution of generated power to identify outliers
plt.figure(figsize=(15,30))
for num,column in enumerate(df.columns):
    if column != 'generated_power_kw':
        plt.subplot(7, 3, num+1)
        plt.boxplot(df[column].dropna(), vert=True, patch_artist=True, 
           boxprops=dict(facecolor='lightblue', alpha=0.7),
           medianprops=dict(color='red', linewidth=2))
        plt.xlabel(column)
# plt.boxplot(df['generated_power_kw'])
# plt.title('Box Plot of Generated Power (KW)')
# plt.xlabel('Generated Power(KW)')
plt.show()
```
