```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#load the dataset
df = pd.read_csv('Iris.csv')
```

```python
# Setting a style similar to Seaborn's 'darkgrid'
plt.style.use('bmh') 
plt.rcParams['figure.figsize'] = (16, 6)

# Create the figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2)

# --- Subplot 1: Boxplot ---
# We use vert=False to make it horizontal like your screenshot
ax1.boxplot(df['petalwidthcm'], vert=False, patch_artist=True, 
            boxprops=dict(facecolor='#2b7bba')) 
ax1.set_title('Boxplot of petalwidthcm')
ax1.set_xlabel('PetalWidthCm')

# --- Subplot 2: Distribution Plot ---
# We use Pandas built-in plotting which handles the Histogram and KDE
# Note: KDE requires 'scipy' to be installed in your JupyterLite environment
df['petalwidthcm'].plot(kind='hist', density=True, bins=5, 
                        alpha=0.5, color='blue', ax=ax2)
df['petalwidthcm'].plot(kind='kde', color='blue', ax=ax2)

ax2.set_title('Distribution of petalwidth')
ax2.set_xlabel('petalwidthcm')
ax2.set_ylabel('Density')

plt.tight_layout()
plt.show()
```
