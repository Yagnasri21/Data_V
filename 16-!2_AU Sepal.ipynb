```python
# Include all the required libraries
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Iris.csv')
```

```python
df.columns
```

```python
plt.figure(figsize=(10,4))
plt.bar(df['SepalLengthCm'].value_counts().index, df['SepalLengthCm'].value_counts().values)
plt.title('sepallength-count')
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.bar(df['SepalLengthCm'].value_counts().index, df['SepalLengthCm'].value_counts().values, data=df)
plt.title("sepallength-count")
```

```python
counts = df['SepalLengthCm'].value_counts().sort_index()
x = counts.index
y = counts.values

# 2. Choose a nice color map (similar to Seaborn palette)
colors = plt.cm.rainbow(np.linspace(0, 1, len(x)))

# 3. Plot
plt.figure(figsize=(10, 4))
plt.bar(x, y, color=colors, edgecolor='black')
plt.title('sepallength-count')
plt.xlabel('sepallengthcm')
plt.ylabel('count')
plt.xticks(x, rotation=90)
plt.tight_layout()
plt.show()
```

```python
df.SepalLengthCm.plot(kind = 'box',color = 'b')
```

```python
plt.figure(figsize=(12,5))
plt.hist([df[df['Species'] == s]['SepalLengthCm'] for s in df['Species'].unique()],
         bins=sorted(df['SepalLengthCm'].unique()),
         stacked=True, label=df['Species'].unique())
```

```python
plt.rcParams['figure.figsize'] = (16, 6)

plt.subplot(1, 2, 1)
plt.boxplot(df['SepalLengthCm'])
plt.title('Boxplot of SepalLengthCm')

plt.subplot(1, 2, 2)
plt.hist(df['SepalLengthCm'], bins=20, color='blue', edgecolor='black', density=False)
plt.title('Distribution of sepallength')

plt.tight_layout()
plt.show()
```

```python
plt.rcParams['figure.figsize'] = (16, 6)

plt.subplot(1,2,1)
plt.boxplot(df['PetalLengthCm'])
plt.title('Boxplot of PetalLengthCm')

plt.subplot(1,2,2)
plt.hist(df['PetalLengthCm'], bins=20, color='black', edgecolor='black', density=True)
plt.title('Distribution of petallength')
plt.xlabel('PetalLengthCm')
plt.ylabel('Density')

plt.tight_layout()
plt.show()
```
